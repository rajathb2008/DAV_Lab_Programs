```python
print("Concatenation")
print("-"*50)
new_sales=pd.DataFrame({
"Order Date":
[
      "2024-06-01",
      "2024-06-05",
      "2024-06-10"
],
"Product name":
[
    "Laptop",
    "Mouse",
    "Printer"
],
"Category":[
    "Electronics",
    "Accessories",
    "Office"
],
"Region":
[
    "North",
    "South",
    "East"
],
"Quantity":
[
    2,
    4,
    3
],
"Sales":
[
    60000,
    2000,
    9000
],
"Profit":
[
    12000,
    400,
    1800
]
})

```

```python
print("New sales Dataframe")
display(new_sales)
```

```python
combined_df=pd.concat(
    [df,new_sales],
    ignore_index=True
)
```

```python
print("Concatenated frame")
print("-"*50)
print(combined_df)
```

```python
product_info=pd.DataFrame({
    "Product name":[
        "Laptop",
        "Mouse",
        "Printer",
        "Tablet",
        "Monitor"
    ],
    "Supplier":
    [
        "Dell",
        "Logitech",
        "HP",
        "Samsung",
        "LG"
    ],
    "Discount":
    [
        10,
        5,
        15,
        8,
        2
    ]
})
```

```python
display(product_info)
```

```python
merged_df=pd.merge(
    df,
    product_info,
    on="Product name",
    how="left"
)
print(merged_df)
```

```python
region_info=pd.DataFrame({
    "Region_Manager":
    [
        "Manager A",
        "Manager B",
        "Manager C",
        "Manager D"
    ],
    "Region":
    [
    "North",
    "South",
    "East",
    "West"
    ]
        })
```

```python
df_region=df.set_index("Region")
joined_df=df_region.join(region_info)
print(joined_df)
```

```python
region_sales=df.groupby(
    "Region"
)["Sales"].sum()
print("Total sales by region")
display(region_sales)
```

```python
pivot_table=pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value=0
)
```

```python
print('Pivot table is\n',pivot_table)
```
